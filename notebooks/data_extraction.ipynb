## Database Connection and Enviroment Setup
**Security protocol:** Credentials are fully externalized to a ignored `.env` file to prevent security leaks.

```python
import os
import pandas as pd
import numpy as np
from sqlalchemy import create_engine
from dotenv import load_dotenv

load_dotenv()
user=os.getenv('user')
database=os.getenv('database')
host=os.getenv('host')
port=os.getenv('port')
password=os.getenv('password')
# format: postgresql:user:password@localhost:port/database
conn=f"postgresql://{user}:{password}@{host}:{port}/{database}"
# Checking Connection
pd.read_sql("select count(*)as total_customers from customers;", conn)
```
