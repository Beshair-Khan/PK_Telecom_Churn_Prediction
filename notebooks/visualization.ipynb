## Loading Data via Reusable Module
* Instead of repeating the CSV loading loop, this imports `load_csv()` from `src/data_loader.py`.
* `sys.path.append("../src")` lets this notebook find that module, since it lives outside `notebooks/`.
* Returns a dictionary of DataFrames, one per CSV in `../data`, keyed by filename (without extension).

```python
import sys
sys.path.append("../src")
from data_loader import load_csv
dfs=load_csv()
dfs['recharge frequency trend'].head()
```

```python
def styled(ax,title="", xlabel="", ylabel=""):
    ax.set_title(title, fontsize=14, fontweight='bold', color='#222222', pad=12, loc='left')
    ax.set_xlabel(xlabel, fontsize=11, fontweight='medium', color='#333333', labelpad=8)
    ax.set_ylabel(ylabel, fontsize=11, fontweight='medium', color='#333333', labelpad=8)
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.spines['bottom'].set_color('#cccccc')
    ax.spines['left'].set_color('#cccccc')
    ax.tick_params(axis='both', labelsize=7, color='#333333') 
    ax.grid(True, axis='both', linestyle='--', alpha=0.5, color='#cccccc')
    ax.set_axisbelow(True) 
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

x=dfs['revenue by operator']['operator']
height=dfs['revenue by operator']['total_revenue']
fig,ax=plt.subplots(figsize=(6,4))
ax.bar(x,height)
styled(ax,title='Revenue By Operator',xlabel='Operator',ylabel='Revenue')
plt.tight_layout()
plt.show()
```

```python
fig,ax=plt.subplots(figsize=(6,4))
operators=dfs['monthly revenue trend']['operator'].unique()
for op in operators:
    op_data=dfs['monthly revenue trend'][dfs['monthly revenue trend']['operator']==op]
    line,=ax.plot(op_data['month'],op_data['total_revenue'],label=op)
    colors=line.get_color()
    ax.fill_between(op_data['month'],op_data['total_revenue'],color=colors,alpha=0.2)
styled(ax,title='Monthly Revenue Trend',xlabel='Month',ylabel='Revenue')
step=6  
all_months=dfs['monthly revenue trend']['month'].unique()
tick_positions = np.arange(0, len(all_months), step)
tick_labels = all_months[tick_positions]
ax.set_xticks(tick_positions)
ax.set_xticklabels(tick_labels, rotation=45, ha='right')
ax.legend(title='Operators', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```
